Ejercicio 1 — Calibrar una lectura

```python
def calibrar_lectura(valor, pendiente=1.0, intercepto=0.0):
  return valor * pendiente + intercepto
sin_calibrar = calibrar_lectura(10.0, 1.2, -0.5)
print("sin calibrar: ", sin_calibrar)
calibrada = calibrar_lectura(10.0, 2.0, 5.0)
print("calibrada: ", calibrada)
con_intercepto = calibrar_lectura(10.0, intercepto=2.0)
print("con intercepto: ", con_intercepto)
```

Ejercicio 2 — ¿Es utilizable la lectura?

```python
lecturas_sbc_a = [10.0, -999.0, 25.0, 40.0, -999.0, 52.0, 600.0]
def es_lectura_sbc_a(valor):
  return valor != -999.0 and (0.0 <= valor <= 500.0)

print("Valida 25.0: ", es_lectura_sbc_a(25.0))
print("Valida -9990.0: ", es_lectura_sbc_a(-999.0))
print("Valida 600.0: ", es_lectura_sbc_a(600.0))
print("Valida 0.0: ", es_lectura_sbc_a(0.0))
```

Ejercicio 3 — Promedio de las lecturas útiles

```python
def promedio_valido(lecturas):
  acumulador = 0.0
  contador = 0
  for valor in lecturas:
    if es_lectura_sbc_a(valor):
      acumulador += valor
      contador += 1
  if contador == 0:
    return None
  return acumulador / contador
promedio_sbc_a = promedio_valido(lecturas_sbc_a)
print("Promedio SBC-A: ", promedio_sbc_a)
print("Promedio lista invalida: ", promedio_valido([10.0, -999.0]))
```

Ejercicio 4 — Calibrar la serie con map, filter y lambda

```python
pendiente_sbc_a = 1.2
intercepto_sbc_a = -0.5

utiles = list(filter(es_lectura_sbc_a, lecturas_sbc_a))
lecturas_calibradas = list(map(lambda x: calibrar_lectura(x, pendiente_sbc_a, intercepto_sbc_a), utiles))
excedencias = len(list(filter(lambda x: x > 35.4, lecturas_calibradas)))

print("Lecturas calibradas:", lecturas_calibradas)
print("Excedencias:", excedencias)
```

Ejercicio 5 — La clase Lectura

```python
class Lectura:
    def __init__(self, valor, unidad="ug/m3", codigo_sensor="sin asignar"):
        self.valor = valor
        self.unidad = unidad
        self.codigo_sensor = codigo_sensor

    def a_microgramos(self):
        if self.unidad == "mg/m3":
            return self.valor * 1000.0
        return self.valor

    def es_dato_faltante(self):
        return self.valor == -999.0

lectura_campo = Lectura(0.0252, "mg/m3", "SBC-A")
lectura_faltante = Lectura(-999.0, codigo_sensor="SBC-A")
lectura_normal = Lectura(30.0)

print("Lectura campo (ug/m3):", lectura_campo.a_microgramos())
print("Lectura faltante es dato faltante:", lectura_faltante.es_dato_faltante())
print("Lectura campo es dato faltante:", lectura_campo.es_dato_faltante())
print("Lectura normal unidad:", lectura_normal.unidad, "| Sensor:", lectura_normal.codigo_sensor)
```
