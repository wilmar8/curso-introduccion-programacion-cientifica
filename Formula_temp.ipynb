Cuando las formas no coinciden: broadcasting

````python
ajuste_dia = np.array([0.5, 0.5, 0.0, -0.5, 0.0, 0.5, 0.0])
````

````python
print(temps + ajuste_dia[0])
````

```markdown
### Explicación del Error de Dimensiones

Al intentar sumar `ajuste_est` de forma `(4,)` directamente a `temps` de forma `(4, 7)`, NumPy alinea las dimensiones desde la derecha:
- `temps`: `(4, 7)`
- `ajuste_est`: `(4,)`

Como `7` no es igual a `4`, NumPy arroja un error de broadcasting.

**Solución (Broadcasting Vertical):**  
Para solucionarlo, debemos transformar el vector a una forma bidimensional `(4, 1)` utilizando `.reshape(4, 1)` o `[:, np.newaxis]`. De este modo:
- `temps`: `(4, 7)`
- `col`: `(4, 1)` -> La dimensión de tamaño `1` se expande automáticamente a `7` para coincidir.
```

````python
ajuste_est = np.array([1.0, 1.0, 0.5, 0.0])
col = ajuste_est.reshape(4, 1)
````

````python
print(temps + col)
````

### Ejercicio de Práctica: Broadcasting Horizontal

Usemos ahora el vector `ajuste_dia` que tiene dimensiones `(7,)` para ajustar cada uno de los 7 días de la semana de todas las estaciones a la vez en la matriz `temps` `(4, 7)`.

````python

temps_con_ajuste_diario = temps + ajuste_dia
print("Temperaturas:", temps.shape)
print("Ajuste dia:", ajuste_dia.shape)
print("Ajuste estacion:", ajuste_est.shape)
print("Temperaturas con ajuste diario:", temps_con_ajuste_diario.shape)
````

````python
col = ajuste_est.reshape(4, 1)
print(col.shape)
print((temps + col)[:, 0])
````
